```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="thegodsangel",
    database="olist_seller_performance"
)

fact_order_items = pd.read_sql("SELECT * FROM fact_order_items;", conn)
fact_order = pd.read_sql("SELECT * FROM fact_order;", conn)
fact_reviews = pd.read_sql("SELECT * FROM fact_reviews;", conn)

print(f"order_items: {len(fact_order_items)} rows")
print(f"orders: {len(fact_order)} rows")
print(f"reviews: {len(fact_reviews)} rows")

df = fact_order_items.merge(fact_order, on="order_id", how="left")
df = df.merge(fact_reviews, on="order_id", how="left")

print(f"Joined: {len(df)} rows, {df['seller_id'].nunique()} unique sellers")

df["is_on_time"] = (
    df["order_delivered_customer_date"].notna() &
    (df["order_delivered_customer_date"] <= df["order_estimated_delivery_date"])
).astype(int)

df["is_failed_order"] = df["order_status"].isin(["canceled", "unavailable"]).astype(int)

scorecard = df.groupby("seller_id").agg(
    order_volume=("order_id", "nunique"),
    avg_review_score=("review_score", "mean"),
    on_time_rate=("is_on_time", "mean"),
    cancellation_rate=("is_failed_order", "mean"),
    total_revenue=("price", "sum")
).reset_index()

MIN_ORDERS = 5
scorecard["eligible"] = scorecard["order_volume"] >= MIN_ORDERS
eligible_df = scorecard[scorecard["eligible"]].copy()
print(f"{len(eligible_df)} of {len(scorecard)} sellers meet the {MIN_ORDERS}+ order threshold")

review_q1 = eligible_df["avg_review_score"].quantile(0.25)
review_q3 = eligible_df["avg_review_score"].quantile(0.75)
ontime_q3 = eligible_df["on_time_rate"].quantile(0.75)
CANCEL_THRESHOLD = 0.10

def assign_tier(row):
    if not row["eligible"]:
        return "Not Rated (low volume)"
    if row["avg_review_score"] >= review_q3 and row["on_time_rate"] >= ontime_q3:
        return "Promote"
    if row["avg_review_score"] <= review_q1 or row["cancellation_rate"] > CANCEL_THRESHOLD:
        return "Warn/Remove"
    return "Monitor"

scorecard["tier"] = scorecard.apply(assign_tier, axis=1)
print("\nTier distribution:")
print(scorecard["tier"].value_counts())

total_orders = df["order_id"].nunique()
warn_sellers = scorecard[scorecard["tier"] == "Warn/Remove"]["seller_id"]
warn_orders = df[df["seller_id"].isin(warn_sellers)]["order_id"].nunique()
warn_order_pct = warn_orders / total_orders * 100

one_star_total = df[df["review_score"] == 1].shape[0]
warn_one_star = df[(df["seller_id"].isin(warn_sellers)) & (df["review_score"] == 1)].shape[0]
warn_one_star_pct = (warn_one_star / one_star_total * 100) if one_star_total > 0 else 0

print(f"\nWarn/Remove sellers: {warn_order_pct:.1f}% of orders, "
      f"but {warn_one_star_pct:.1f}% of all 1-star reviews")

cursor = conn.cursor()
cursor.execute("DROP TABLE IF EXISTS dim_seller_scorecard;")
cursor.execute("""
    CREATE TABLE dim_seller_scorecard (
        seller_id VARCHAR(50) PRIMARY KEY,
        order_volume INT,
        avg_review_score FLOAT,
        on_time_rate FLOAT,
        cancellation_rate FLOAT,
        total_revenue FLOAT,
        eligible BOOLEAN,
        tier VARCHAR(30)
    );
""")
conn.commit()

insert_query = """
    INSERT INTO dim_seller_scorecard 
    (seller_id, order_volume, avg_review_score, on_time_rate, cancellation_rate, total_revenue, eligible, tier)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
"""
data = scorecard[["seller_id", "order_volume", "avg_review_score", "on_time_rate",
                   "cancellation_rate", "total_revenue", "eligible", "tier"]].values.tolist()
cursor.executemany(insert_query, data)
conn.commit()

print(f"\ndim_seller_scorecard written: {cursor.rowcount} rows")
cursor.close()
conn.close()
```
